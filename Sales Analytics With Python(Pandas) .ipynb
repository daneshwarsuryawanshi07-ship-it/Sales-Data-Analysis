```python
df = df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
product_analysis = (
    df.groupby("Product")[["Sales", "Profit"]]
    .sum()
    .sort_values("Sales",ascending=False)
)
print(product_analysis.head(10))
```

```python
city_anaylsis = (
    df.groupby("City")["Sales"]
    .sum()
    .sort_values(ascending=False)
)
print(city_anaylsis)
```

```python
customer_segment_analysis = (
    df.groupby("Customer Segment") [["Sales", "Profit"]]
    .sum()
)
print(customer_segment_analysis)
```
